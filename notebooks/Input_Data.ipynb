## HanLP + Stanza SRL Parse

Use this section to run HanLP semantic role labelling and Stanza component lemmatisation into SynFlow's seven-field format.


```python
import hanlp

srl_input_root = Path("/home/volt/bach/pilot_data/float_gen")
srl_output_root = Path("/home/volt/bach/pilot_data/float_gen_srl")

# GPUs and parallel processing.
srl_gpu = "2,3"
srl_workers_per_gpu = 1

# HanLP sentence batch size.
hanlp_batch_size = 32

# Choose the HanLP SRL model and Stanza lemmatisation model explicitly.
hanlp_srl_model_name = (
    hanlp.pretrained.mtl.EN_TOK_LEM_POS_NER_SRL_UDEP_SDP_CON_MODERNBERT_LARGE
)
stanza_srl_language = "en"
stanza_srl_package = {"tokenize": "ewt", "pos": "ewt", "lemma": "ewt"}
```

```python
# If this is the first time using this Stanza language package, run once:
# import stanza
# stanza.download(stanza_srl_language)

hanlp_stanza_parse_folder(
    srl_input_root,
    srl_output_root,
    hanlp_model_name=hanlp_srl_model_name,
    language=stanza_srl_language,
    stanza_package=stanza_srl_package,
    gpu=srl_gpu,
    workers_per_gpu=srl_workers_per_gpu,
    hanlp_batch_size=hanlp_batch_size,
    overwrite=False,
)
```

## FrameSemanticTransformer + Stanza Frame Parse

Use this section to run FrameSemanticTransformer frame parsing and Stanza component lemmatisation into SynFlow's seven-field format.


```python
frame_input_root = Path("/home/volt/bach/pilot_data/float_gen")
frame_output_root = Path("/home/volt/bach/pilot_data/float_gen_frames")

# GPUs and parallel processing.
frame_gpu = "2,0"
frame_workers_per_gpu = 1

# Choose the FrameSemanticTransformer and Stanza lemmatisation models explicitly.
frame_model_name = "base"
small_frame_model_name = "small"
frame_batch_size = 32
use_small_frame_fallback = True
stanza_frame_language = "en"
stanza_frame_package = {"tokenize": "ewt", "pos": "ewt", "lemma": "ewt"}
```

```python
# If this is the first time using this Stanza language package, run once:
# import stanza
# stanza.download(stanza_frame_language)

frame_stanza_parse_folder(
    frame_input_root,
    frame_output_root,
    frame_model_name=frame_model_name,
    small_frame_model_name=small_frame_model_name,
    frame_batch_size=frame_batch_size,
    use_small_fallback=use_small_frame_fallback,
    language=stanza_frame_language,
    stanza_package=stanza_frame_package,
    gpu=frame_gpu,
    workers_per_gpu=frame_workers_per_gpu,
    overwrite=False,
)
```
