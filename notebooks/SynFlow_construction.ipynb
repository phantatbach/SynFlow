```python
# Specify required input

# Add SynFlow to path in order to import modules
repo_root = "/home/volt/bach/SynFlow"
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

# Regex corpus_pattern to extract relevant information from CoNLL-U files
corpus_pattern = re.compile(
    r"([^\t]+)\t"      # word form
    r"([^\t]+)\t"      # lemma
    r"([^\t]+)\t"      # POS (UPOS or XPOS)
    r"([^\t]+)\t"      # ID
    r"([^\t]+)\t"      # HEAD
    r"([^\t]+)"        # DEPREL
)

# Specify input
target_construction = "ditransitive"
target_construction_template = "(VERB) & > (chi_nsubj, []) & > (chi_obj, []) & > (chi_iobj, [])"

# Specify corpus and output folders
period = "1810-2000"
corpus_folder = f"/home/volt/bach/pilot_data/1810s/"
output_folder = Path(f"/home/volt/bach/SynFlow/case_studies/SynFlow-merge/")
```
